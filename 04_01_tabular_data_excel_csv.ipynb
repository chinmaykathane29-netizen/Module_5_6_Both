## 1. Why data management matters (10 min)

A plant may record thousands of sensor values, laboratory results, and batch records. Useful analysis depends on data being understandable and traceable.

### A tidy-table habit
- One **row** = one observation (for example, one hourly reading).
- One **column** = one variable (for example, temperature in °C).
- One **cell** = one value.
- Give columns meaningful names including units: `flow_m3_h`, not `Flow1`.

Excel is convenient for manual inspection; **CSV** is a plain-text table format that moves easily between Excel, Python, and reporting tools.

## 3. Inspect before analysing (15 min)

Never jump directly to a conclusion from a new file. First inspect its size, column names, data types, and a few rows. `head()` gives the first rows; `info()` describes columns; `describe()` gives numerical summaries.

```python
print('Shape (rows, columns):', process_data.shape)
print('\nColumn names:', list(process_data.columns))
display(process_data.head())
process_data.info()
display(process_data.describe())
```

## 4. Excel and CSV: practical workflow (10 min)

A recommended workflow for this course:

1. Collect or export a controlled data table from a spreadsheet/system.
2. Save a working copy as **CSV UTF-8** with a clear file name and units in headers.
3. Import to Python, inspect it, clean it, then create repeatable outputs.
4. Keep the raw source unchanged; save cleaned data separately.

Avoid merged cells, decorative title rows, and mixed units in a single numeric column: they make reliable analysis harder.
